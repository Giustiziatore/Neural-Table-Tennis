```python
# --- STEP 0: INSTALLAZIONI E IMPORT ---
# !pip install mediapipe opencv-python seaborn scikit-learn tensorflow

import cv2
import mediapipe as mp
import numpy as np
import os
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report
from tensorflow.keras import layers, models, optimizers, callbacks
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
from tensorflow.keras.models import load_model
```

```python
# --- CONFIGURAZIONE DATASET E MODELLO ---
# IMPORTANTE: Se usi Kaggle, cambia questo percorso con quello del tuo dataset caricato!
# Se sei sul tuo PC, metti il percorso della cartella che contiene "regular" e "irregular"
dataset_path = "dataset" 

CLASSES = ["irregolari", "regolari"] # Label 0 e Label 1
MAX_FRAMES = 90 # 3 secondi a 30fps (Standardizzazione lunghezza video)

checkpoint = ModelCheckpoint('best_model.keras', monitor='val_accuracy', save_best_only=True, mode='max', verbose=1)
early_stopping = EarlyStopping(monitor='val_loss', patience=25, restore_best_weights=True)
```

```python
# --- STEP 1: FUNZIONI DI ESTRAZIONE (MEDIAPIPE) ---
mp_pose = mp.solutions.pose
pose = mp_pose.Pose(
    static_image_mode=False, #False per video
    model_complexity=1, #complessità del modello 0, 1, 2
    min_detection_confidence=0.5, #soglia di rilevamento
    min_tracking_confidence=0.5 #soglia di tracciamento
)
```

```python
def estrai_feature(video_path):
    cap = cv2.VideoCapture(video_path)
    frames_data = []
    
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
            
        # Converti in RGB per MediaPipe
        image_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        results = pose.process(image_rgb)
        
        if results.pose_landmarks:
            row = []
            for landmark in results.pose_landmarks.landmark:
                # Prendiamo x, y, z e visibility (33 punti * 4 valori = 132 features)
                row.extend([landmark.x, landmark.y, landmark.z, landmark.visibility])
            frames_data.append(row)
        else:
            # Se MediaPipe perde la persona per un frame, mettiamo zeri o saltiamo
            # Qui mettiamo una riga di zeri per mantenere la continuità temporale
            frames_data.append([0] * 132)
            
    cap.release()
    return np.array(frames_data)
```

```python
def carica_dataset(base_path):
    X = []
    y = []
    
    print(f"Inizio caricamento da: {base_path}")
    
    for label_idx, class_name in enumerate(CLASSES):
        folder_path = os.path.join(base_path, class_name)
        if not os.path.exists(folder_path):
            print(f"ERRORE: La cartella {folder_path} non esiste!")
            continue
            
        video_files = [f for f in os.listdir(folder_path) if f.endswith(('.mp4', '.mov', '.avi', '.mkv'))]
        print(f"  -> Trovati {len(video_files)} video in '{class_name}'")
        
        for video_file in video_files:
            full_path = os.path.join(folder_path, video_file)
            
            # Estrazione Features
            features = estrai_feature(full_path)
            
            # Padding / Truncating (Per avere tutti array lunghi 90)
            if len(features) == 0: continue # Salta video vuoti
            
            if len(features) > MAX_FRAMES:
                features = features[:MAX_FRAMES] # Taglia se troppo lungo
            elif len(features) < MAX_FRAMES:
                # Aggiungi zeri se troppo corto
                diff = MAX_FRAMES - len(features)
                padding = np.zeros((diff, 132))
                features = np.concatenate((features, padding))
                
            X.append(features)
            y.append(label_idx)
            
    return np.array(X), np.array(y)
```

```python
def data_augmentation(X, y):
    X_aug = []
    y_aug = []
    
    for i in range(len(X)):
        # 1. Aggiungi l'originale
        original_sample = X[i]
        X_aug.append(original_sample)
        y_aug.append(y[i])
        
        # 2. Crea il gemello specchiato
        # Copia il campione
        mirrored_sample = original_sample.copy()
        
        # In MediaPipe i dati sono [x, y, z, vis, x, y, z, vis, ...]
        # La X è ogni 4 posizioni: indici 0, 4, 8, 12...
        # Per specchiare orizzontalmente (flip): nuovo_x = 1.0 - vecchio_x
        # (Nota: ignoriamo le righe di padding che sono tutte zeri)
        
        # Iteriamo su tutti i frame (90)
        for frame_idx in range(len(mirrored_sample)):
            # Se è tutto zero (padding), salta
            if np.sum(mirrored_sample[frame_idx]) == 0:
                continue
                
            # Modifica solo le X (colonne 0, 4, 8... fino a 132)
            for feature_idx in range(0, 132, 4):
                mirrored_sample[frame_idx][feature_idx] = 1.0 - mirrored_sample[frame_idx][feature_idx]
        
        X_aug.append(mirrored_sample)
        y_aug.append(y[i])
        
    return np.array(X_aug), np.array(y_aug)


```

```python
from sklearn.model_selection import train_test_split

# --- STEP 2: CARICAMENTO DATI GREZZI ---
# Carichiamo solo gli originali (62 video circa)
X_raw, y_raw = carica_dataset(dataset_path)

print(f"Dati ORIGINALI caricati: {X_raw.shape}") 
# (Dovrebbe essere circa 62 campioni)

# --- STEP 3: SPLIT PRIMA DELL'AUGMENTATION ---
# Dividiamo subito. Qui decidiamo quali video il modello NON vedrà mai.
# Usa stratify=y_raw per mantenere le proporzioni giuste tra regolari/irregolari
X_train_raw, X_test_raw, y_train_raw, y_test_raw = train_test_split(
    X_raw, y_raw, test_size=0.2, random_state=42, stratify=y_raw
)

print(f"\n--- DIVISIONE EFFETTUATA ---")
print(f"Video nel Train (Originali): {len(X_train_raw)}")
print(f"Video nel Test (Originali): {len(X_test_raw)}")

# --- STEP 4: DATA AUGMENTATION (SEPARATA) ---
print(f"\n--- APPLICAZIONE AUGMENTATION ---")

# 1. Augmentiamo il Training Set (Fondamentale per raddoppiare i dati di studio)
X_train, y_train = data_augmentation(X_train_raw, y_train_raw)
print(f"Training Set Finale (Originali + Specchiati): {X_train.shape}")

# 2. Aumentiamo anche il Test Set
# (da 13 diventano 26 test),
# Derivati SOLO da X_test_raw!
X_test, y_test = data_augmentation(X_test_raw, y_test_raw)
print(f"Test Set Finale (Originali + Specchiati): {X_test.shape}")

# Ora X_train e X_test sono pronti per il modello
```

```python
# --- STEP 4: DEFINIZIONE MODELLO  ---
model = models.Sequential([
    # Masking ignora gli zeri del padding (fondamentale per video corti)
    layers.Masking(mask_value=0.0, input_shape=(MAX_FRAMES, 132)),

    # LSTM Layer: Impara la sequenza temporale
    layers.LSTM(64, return_sequences=False, dropout=0.2, recurrent_dropout=0.2),

    # Dense Layers: Classificazione
    layers.Dense(32, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(1, activation='sigmoid') # Output 0 o 1
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# --- STEP 5: ADDESTRAMENTO ---
print("\n--- INIZIO TRAINING ---")
history = model.fit(
    X_train, y_train,
    epochs=100,
    callbacks=[checkpoint, early_stopping],          
    batch_size=4,       # Piccolo batch perché abbiamo pochi dati
    validation_split=0.2, # Usa una parte del train per validare mentre impara
    verbose=1
)
```

```python
# --- STEP 6: VALUTAZIONE E GRAFICI ---
# Calcola accuracy finale sul test set (quello mai visto)
test_loss, test_acc = model.evaluate(X_test, y_test)
print(f"\nACCURACY FINALE SUL TEST SET: {test_acc*100:.2f}%")

# Plot Grafici
plt.figure(figsize=(12, 5))

# Grafico Accuracy
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Train Acc')
plt.plot(history.history['val_accuracy'], label='Val Acc')
plt.title('Andamento Accuracy')
plt.legend()

# Grafico Loss
plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.title('Andamento Errore (Loss)')
plt.legend()
plt.show()

# Matrice di Confusione
y_pred = (model.predict(X_test) > 0.5).astype(int)
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=CLASSES, yticklabels=CLASSES)
plt.title('Matrice di Confusione (Test Set)')
plt.ylabel('Reale')
plt.xlabel('Predetto dall\'AI')
plt.show()
```

```python
# --- STEP 8: INFERENZA SU VIDEO ESTERNI ---
from tensorflow.keras.models import load_model

# --- CONFIGURAZIONE ---
# Metti qui il percorso del tuo video da testare
VIDEO_PATH = "test/video_inferenza.mp4" 
MODEL_PATH = "best_model.keras" # Usiamo il modello migliore salvato
MAX_FRAMES = 90 # Deve essere LO STESSO usato in addestramento

# --- 2. FUNZIONE DI PREPARAZIONE ---
def prepare_video(video_path):
    print(f"Analisi del video: {video_path}...")
    cap = cv2.VideoCapture(video_path)
    frames_data = []
    
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
            
        # Conversione e MediaPipe
        image_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        results = pose.process(image_rgb)
        
        if results.pose_landmarks:
            row = []
            for landmark in results.pose_landmarks.landmark:
                row.extend([landmark.x, landmark.y, landmark.z, landmark.visibility])
            frames_data.append(row)
        else:
            # Se non vede nessuno, mette zeri (come in training)
            frames_data.append([0] * 132)
            
    cap.release()
    
    # --- PADDING (Cruciale!) ---
    # Dobbiamo trasformare questa lista in una matrice (1, 90, 132)
    # L'AI si aspetta ESATTAMENTE quella forma.
    
    data = np.array(frames_data)
    
    # Se il video è vuoto o non ha trovato nessuno
    if len(data) == 0:
        print("Errore: Nessuna persona rilevata nel video.")
        return None

    # Taglia o Allunga a 90 frame
    if len(data) > MAX_FRAMES:
        data = data[:MAX_FRAMES] # Taglia
    elif len(data) < MAX_FRAMES:
        diff = MAX_FRAMES - len(data)
        padding = np.zeros((diff, 132)) # Aggiungi zeri
        data = np.concatenate((data, padding))
        
    # Aggiungi la dimensione del "Batch"
    # L'AI non vuole vedere (90, 132), vuole vedere (1, 90, 132)
    # Cioè "Un pacchetto contenente un video"
    data = np.expand_dims(data, axis=0)
    
    return data

# --- 3. ESECUZIONE ---

# A. Carica il modello
print("Caricamento del cervello...")
model = load_model(MODEL_PATH)

# B. Prepara i dati
input_data = prepare_video(VIDEO_PATH)

if input_data is not None:
    # C. Predizione
    prediction = model.predict(input_data)
    score = prediction[0][0] # Prende il numero singolo (tra 0 e 1)

    # D. Risultato
    print("\n----------------RISULTATO----------------")
    print(f"Punteggio AI: {score:.4f}")
    
    if score > 0.5:
        print(f" REGOLARE (Confidenza: {score*100:.1f}%)")
    else:
        print(f" IRREGOLARE (Confidenza: {(1-score)*100:.1f}%)")
    print("-----------------------------------------")
```

```python
from sklearn.model_selection import StratifiedKFold
import numpy as np

# Configurazione
NUM_FOLDS = 5
BATCH_SIZE = 4
EPOCHS = 50

# Usa StratifiedKFold invece di KFold semplice (meglio per dataset piccoli)
kfold = StratifiedKFold(n_splits=NUM_FOLDS, shuffle=True, random_state=42)

acc_per_fold = []
loss_per_fold = []

fold_no = 1
print(f"--- INIZIO CROSS VALIDATION ({NUM_FOLDS} Folds) ---")

# IMPORTANTE: Splitta su X_raw e y_raw (i dati originali non aumentati)
for train_index, test_index in kfold.split(X_raw, y_raw):
    
    # --- 1. Dividiamo i dati RAW per questo giro ---
    X_train_raw_fold = X_raw[train_index]
    X_test_raw_fold  = X_raw[test_index]
    y_train_raw_fold = y_raw[train_index]
    y_test_raw_fold  = y_raw[test_index]
    
    # --- 2. APPLICHIAMO DATA AUGMENTATION QUI (DENTRO IL LOOP) ---
    # Fondamentale: aumentiamo i dati di questo specifico fold
    X_train_fold, y_train_fold = data_augmentation(X_train_raw_fold, y_train_raw_fold)
    
    # Opzionale: Aumentiamo anche il test per avere più casi di verifica
    X_test_fold, y_test_fold = data_augmentation(X_test_raw_fold, y_test_raw_fold)
    
    # Debug dimensionale (opzionale, per vedere che funziona)
    # print(f"Fold {fold_no} - Train size: {len(X_train_fold)}, Test size: {len(X_test_fold)}")

    # --- 3. Ricostruiamo il modello da ZERO ---
    model = models.Sequential([
        layers.Masking(mask_value=0.0, input_shape=(MAX_FRAMES, 132)),
        layers.GRU(64, return_sequences=False, dropout=0.2),
        layers.Dense(32, activation='relu'),
        layers.Dropout(0.3),
        layers.Dense(1, activation='sigmoid')
    ])
    model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
    
    print(f"\nTraining Fold {fold_no}...")
    
    # --- 4. Addestramento sui dati AUMENTATI ---
    history = model.fit(
        X_train_fold, y_train_fold,
        batch_size=BATCH_SIZE,
        epochs=EPOCHS,
        verbose=0 # Silenzioso
    )
    
    # --- 5. Valutazione ---
    scores = model.evaluate(X_test_fold, y_test_fold, verbose=0)
    print(f"Score Fold {fold_no}: Accuracy {scores[1]*100:.2f}%")
    
    acc_per_fold.append(scores[1] * 100)
    loss_per_fold.append(scores[0])
    
    # Salvataggio del migliore del gruppo
    if scores[1] * 100 >= max(acc_per_fold):
        model.save(f'best_model_fold_{fold_no}.keras')
        # print(" -> Record salvato")
        
    fold_no += 1

# --- RISULTATI FINALI ---
print("\n----------------------------------------------------------------")
print("RISULTATI CROSS-VALIDATION (Corretti con Augmentation)")
print("----------------------------------------------------------------")
print(f"Lista Accuracy: {acc_per_fold}")
print(f"ACCURACY MEDIA: {np.mean(acc_per_fold):.2f}% (+/- {np.std(acc_per_fold):.2f}%)")
print("----------------------------------------------------------------")
```

```python
# --- STEP 9: REPORT DI CLASSIFICAZIONE ---
# y_test sono le risposte vere (0 o 1)
# y_pred sono le risposte dell'AI (convertite in 0 o 1)
y_pred_probs = model.predict(X_test)
y_pred = (y_pred_probs > 0.5).astype(int)

print(classification_report(y_test, y_pred, target_names=["Irregolare", "Regolare"]))
```
