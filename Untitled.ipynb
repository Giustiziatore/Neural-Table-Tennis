```python
import cv2
import mediapipe as mp
import numpy as np
import os
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report
from tensorflow.keras import layers, models, optimizers, callbacks, regularizers
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
from tensorflow.keras.models import load_model
```

<font color='blue'><h5> <center> --- CONFIGURAZIONE DATASET ---  </center>     </h5></font>

```python
dataset_path = "dataset" 

CLASSES = ["irregolari", "regolari"]  # Label 0 e Label 1
MAX_FRAMES = 90 # 3 secondi a 30fps (Standardizzazione lunghezza video)
 
```

<font color='blue'><h5> <center> # --- STEP 1: FUNZIONI DI ESTRAZIONE (MEDIAPIPE) ---     </center>     </h5></font>

```python
mp_pose = mp.solutions.pose
pose = mp_pose.Pose(
    static_image_mode=False, #False per video
    model_complexity=1, #complessità del modello 0, 1, 2
    min_detection_confidence=0.5, #soglia di rilevamento
    min_tracking_confidence=0.5 #soglia di tracciamento
)
```

```python
def estrai_feature(video_path):
    cap = cv2.VideoCapture(video_path)
    frames_data = []
    
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
            
        # Conversione in RGB per MediaPipe
        image_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        results = pose.process(image_rgb)
        
        if results.pose_landmarks:
            row = []
            for landmark in results.pose_landmarks.landmark:
                # x, y, z e visibility (33 punti * 4 valori = 132 features)
                row.extend([landmark.x, landmark.y, landmark.z, landmark.visibility])
            frames_data.append(row)
        else:
            # Se MediaPipe perde la persona per un frame, mettiamo zeri 
            frames_data.append([0] * 132)
            
    cap.release()
    return np.array(frames_data)
```

<font color='blue'><h5> <center> ---   FUNZIONE CARICAMENTO DATASET ---  </center>     </h5></font>

```python
def carica_dataset(base_path):
    X = []
    y = []
    
    print(f"Inizio caricamento da: {base_path}")
    
    for label_idx, class_name in enumerate(CLASSES):
        folder_path = os.path.join(base_path, class_name)
        if not os.path.exists(folder_path):
            print(f"ERRORE: La cartella {folder_path} non esiste!")
            continue
            
        video_files = [f for f in os.listdir(folder_path) if f.endswith(('.mp4', '.mov', '.avi', '.mkv'))]
        print(f"  -> Trovati {len(video_files)} video in '{class_name}'")
        
        for video_file in video_files:
            full_path = os.path.join(folder_path, video_file)
            
            # Estrazione Features
            features = estrai_feature(full_path)
            
            # Padding / Truncating (Per avere tutti array lunghi 90)
            if len(features) == 0: continue # Salta video vuoti
            
            if len(features) > MAX_FRAMES:
                features = features[:MAX_FRAMES] # Taglia se troppo lungo
            elif len(features) < MAX_FRAMES:
                # Aggiungi zeri se troppo corto
                diff = MAX_FRAMES - len(features)
                padding = np.zeros((diff, 132))
                features = np.concatenate((features, padding))
                
            X.append(features)
            y.append(label_idx)
            
    return np.array(X), np.array(y)
```

<font color='blue'><h5> <center> ---    FUNZIONE DATA AUGMENTATION  ---  </center>     </h5></font>

```python
def data_augmentation(X, y):
    X_aug = []
    y_aug = []
    
    for i in range(len(X)):
        # 1. Aggiungi l'originale
        original_sample = X[i]
        X_aug.append(original_sample)
        y_aug.append(y[i])
        # 2. Crea la versione specchiata orizzontalmente
        mirrored_sample = original_sample.copy()
        
        # In MediaPipe i dati sono [x, y, z, vis, x, y, z, vis, ...]
        # La X è ogni 4 posizioni: indici 0, 4, 8, 12...
        # orizzontal flip: nuovo_x = 1.0 - vecchio_x
        
        # Iteriamo su tutti i frame (90)
        for frame_idx in range(len(mirrored_sample)):
            # Se è tutto zero (padding), salta
            if np.sum(mirrored_sample[frame_idx]) == 0:
                continue
                
            # Modifica solo le X (colonne 0, 4, 8... fino a 132)
            for feature_idx in range(0, 132, 4):
                mirrored_sample[frame_idx][feature_idx] = 1.0 - mirrored_sample[frame_idx][feature_idx]
        
        X_aug.append(mirrored_sample)
        y_aug.append(y[i])
        
    return np.array(X_aug), np.array(y_aug)


```

```python
from sklearn.model_selection import train_test_split

X_raw, y_raw = carica_dataset(dataset_path)

print(f"Dati ORIGINALI caricati: {X_raw.shape}") 

X_train_raw, X_test_raw, y_train_raw, y_test_raw = train_test_split(
    X_raw, y_raw, test_size=0.2, random_state=42, stratify=y_raw
)

print(f"\n--- DIVISIONE EFFETTUATA ---")
print(f"Video nel Train (Originali): {len(X_train_raw)}")
print(f"Video nel Test (Originali): {len(X_test_raw)}")
```


<font color='blue'><h5> <center> --- STEP 4: DATA AUGMENTATION (SEPARATA) ---   </center>     </h5></font>
 <center>1. Aumentiamo prima il Training Set </center>
 <center>2. Aumentiamo anche il Test Set solo successivamente alla separazione precedente</center>

```python
print(f"\n--- APPLICAZIONE AUGMENTATION ---")

X_train, y_train = data_augmentation(X_train_raw, y_train_raw)
print(f"Training Set Finale (Originali + Specchiati): {X_train.shape}")


X_test, y_test = data_augmentation(X_test_raw, y_test_raw)
print(f"Test Set Finale (Originali + Specchiati): {X_test.shape}")
```

<font color='blue'><h5> <center>  --- STEP 5: DEFINIZIONE MODELLO  ---   </center>     </h5></font>  

```python
checkpoint = ModelCheckpoint('best_model.keras', monitor='val_accuracy', save_best_only=True, mode='max', verbose=1)
early_stopping = EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True)
reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=5, min_lr=1e-5, verbose=1)  

model = models.Sequential([
    # Masking padding (per ignorare gli zeri dei video corti)
    layers.Masking(mask_value=0.0, input_shape=(MAX_FRAMES, 132)),

    layers.LSTM(32, return_sequences=False, dropout=0.2, recurrent_dropout=0.2),
    layers.Dense(16, activation='relu'),
    layers.Dense(1, activation='sigmoid') # Output 0 o 1
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

<font color='blue'><h5> <center> --- STEP 6: ADDESTRAMENTO --- </center>     </h5></font> 

```python

print("\n--- INIZIO TRAINING ---")
history = model.fit(
    X_train, y_train,
    epochs=50,
    callbacks=[checkpoint , early_stopping, reduce_lr],          
    batch_size=4,       # Piccolo batch perché abbiamo pochi dati (di default 32)
    validation_split=0.2, # Parte del train per validare durante l'addestramento
    verbose=1
)
```

<font color='blue'><h5> <center> --- STEP 6: VALUTAZIONE E GRAFICI --- </center>     </h5></font> 



```python
# Accuracy finale sul test set 
test_loss, test_acc = model.evaluate(X_test, y_test)
print(f"\nACCURACY FINALE SUL TEST SET: {test_acc*100:.2f}%")

# Plot Grafici
plt.figure(figsize=(12, 5))

# Grafico Accuracy
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Train Acc')
plt.plot(history.history['val_accuracy'], label='Val Acc')
plt.title('Andamento Accuracy')
plt.legend()

# Grafico Loss
plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.title('Andamento Errore (Loss)')
plt.legend()
plt.show()

# Matrice di Confusione
y_pred = (model.predict(X_test) > 0.5).astype(int)
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=CLASSES, yticklabels=CLASSES)
plt.title('Matrice di Confusione (Test Set)')
plt.ylabel('Reale')
plt.xlabel('Predetto dall\'AI')
plt.show()
```

# --- STEP 8: INFERENZA SU VIDEO ESTERNI ---

```python

from tensorflow.keras.models import load_model

# --- CONFIGURAZIONE ---
VIDEO_PATH = "test/video_inferenza3.mp4" 
MODEL_PATH = "best_model_fold_2.keras" 
MAX_FRAMES = 90 

# --- 2. FUNZIONE DI PREPARAZIONE ---
def prepare_video(video_path):
    print(f"Analisi del video: {video_path}...")
    cap = cv2.VideoCapture(video_path)
    frames_data = []
    
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
            
        # Conversione e MediaPipe
        image_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        results = pose.process(image_rgb)
        
        if results.pose_landmarks:
            row = []
            for landmark in results.pose_landmarks.landmark:
                row.extend([landmark.x, landmark.y, landmark.z, landmark.visibility])
            frames_data.append(row)
        else:
            frames_data.append([0] * 132)
            
    cap.release()
    
    # --- PADDING---
    # Da (90, 132) a (1, 90, 132) da lista  a matrice
    
    data = np.array(frames_data)
    
    # Se il video è vuoto o non ha trovato nessuno
    if len(data) == 0:
        print("Errore: Nessuna persona rilevata nel video.")
        return None

    # Taglia o Allunga a 90 frame
    if len(data) > MAX_FRAMES:
        data = data[:MAX_FRAMES] # Taglia
    elif len(data) < MAX_FRAMES:
        diff = MAX_FRAMES - len(data)
        padding = np.zeros((diff, 132)) # Aggiungi zeri
        data = np.concatenate((data, padding))
        
    # funzione numpy espandi dimensione batch
    data = np.expand_dims(data, axis=0)
    
    return data

# --- 3. ESECUZIONE ---

# A. Carica il modello
print("Caricamento del cervello...")
model = load_model(MODEL_PATH)

# B. Prepara i dati
input_data = prepare_video(VIDEO_PATH)

if input_data is not None:
    # C. Predizione
    prediction = model.predict(input_data)
    score = prediction[0][0] # Prende il numero singolo (tra 0 e 1)

    # D. Risultato
    print("\n----------------RISULTATO----------------")
    print(f"Punteggio AI: {score:.4f}")
    
    if score > 0.5:
        print(f" REGOLARE (Confidenza: {score*100:.1f}%)")
    else:
        print(f" IRREGOLARE (Confidenza: {(1-score)*100:.1f}%)")
    print("-----------------------------------------")
```

```python
from sklearn.model_selection import StratifiedKFold
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras import models, layers


# Configurazione
NUM_FOLDS = 5
BATCH_SIZE = 4
EPOCHS = 50

# StratifiedKFold
kfold = StratifiedKFold(n_splits=NUM_FOLDS, shuffle=True, random_state=42)

acc_per_fold = []
loss_per_fold = []

fold_no = 1
print(f"--- INIZIO CROSS VALIDATION ({NUM_FOLDS} Folds) ---")

for train_index, test_index in kfold.split(X_, y_):
    
    # --- 1. Split dei dati grezzi ---
    X_train_raw_fold = X_[train_index]
    X_test_raw_fold  = X_[test_index]
    y_train_raw_fold = y_[train_index]
    y_test_raw_fold  = y_[test_index]

    # --- 2. DATA AUGMENTATION ---
    X_train_fold, y_train_fold = data_augmentation(X_train_raw_fold, y_train_raw_fold)
    # Nota: Aumentiamo anche il test per usarlo come validazione durante il training
    X_test_fold, y_test_fold = data_augmentation(X_test_raw_fold, y_test_raw_fold)

    # --- 3. Ricostruzione Modello ---
    model = models.Sequential([
        layers.Masking(mask_value=0.0, input_shape=(MAX_FRAMES, 132)),
        layers.LSTM(32, return_sequences=False, dropout=0.2),
        layers.Dense(16, activation='relu'),
        layers.Dense(1, activation='sigmoid')
    ])
    model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
    
    print(f"\nTraining Fold {fold_no}...")
    
    # --- 4. Addestramento ---
    history = model.fit(
        X_train_fold, y_train_fold,
        validation_data=(X_test_fold, y_test_fold), 
        batch_size=BATCH_SIZE,
        epochs=EPOCHS,
        verbose=0 
    )
    
    # --- 5. Valutazione Finale ---
    scores = model.evaluate(X_test_fold, y_test_fold, verbose=0)
    print(f"-> Score Fold {fold_no}: Accuracy {scores[1]*100:.2f}% | Loss {scores[0]:.4f}")
    
    acc_per_fold.append(scores[1] * 100)
    loss_per_fold.append(scores[0])

    # --- 6. PLOT DEI GRAFICI PER IL FOLD CORRENTE ---
    plt.figure(figsize=(12, 4))
    
    # Grafico Accuracy
    plt.subplot(1, 2, 1)
    plt.plot(history.history['accuracy'], label='Train Accuracy', color='blue')
    plt.plot(history.history['val_accuracy'], label='Val Accuracy', color='orange')
    plt.title(f'Fold {fold_no} - Accuracy')
    plt.ylabel('Accuracy')
    plt.xlabel('Epoch')
    plt.legend()
    plt.grid(True, alpha=0.3)

    # Grafico Loss
    plt.subplot(1, 2, 2)
    plt.plot(history.history['loss'], label='Train Loss', color='blue')
    plt.plot(history.history['val_loss'], label='Val Loss', color='orange')
    plt.title(f'Fold {fold_no} - Loss')
    plt.ylabel('Loss')
    plt.xlabel('Epoch')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show() # Mostra i grafici prima di passare al prossimo fold

    # Salvataggio del migliore
    if scores[1] * 100 >= max(acc_per_fold):
        model.save(f'best_model_fold_{fold_no}.keras')
        
    fold_no += 1

# --- RISULTATI FINALI ---
print("\n" + "="*60)
print("RISULTATI FINALI CROSS-VALIDATION")
print("="*60)
print(f"Lista Accuracy per fold: {[f'{x:.2f}%' for x in acc_per_fold]}")
print(f"ACCURACY MEDIA: {np.mean(acc_per_fold):.2f}% (+/- {np.std(acc_per_fold):.2f}%)")
print(f"LOSS MEDIA:     {np.mean(loss_per_fold):.4f}")
print("="*60)

# Grafico riassuntivo finale (Bar Plot)
plt.figure(figsize=(8, 5))
plt.bar(range(1, NUM_FOLDS+1), acc_per_fold, color='skyblue', edgecolor='black')
plt.axhline(y=np.mean(acc_per_fold), color='r', linestyle='--', label=f'Media: {np.mean(acc_per_fold):.1f}%')
plt.xlabel('Fold Number')
plt.ylabel('Accuracy (%)')
plt.title('Accuracy per ogni Fold')
plt.legend()
plt.ylim(0, 100)
plt.show()
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models, regularizers
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import numpy as np

# --- 1. DEFINIZIONE DEL NUOVO MODELLO (CNN 1D) ---
print("Creazione Modello CNN 1D...")

model = models.Sequential([
    # Input Masking (Ignora i frame vuoti)
    layers.Masking(mask_value=0.0, input_shape=(MAX_FRAMES, 132)),
    
    # Primo blocco Convoluzionale (Cerca pattern locali)
    layers.Conv1D(filters=32, kernel_size=3, activation='relu', padding='same'),
    layers.BatchNormalization(), # Stabilizza l'apprendimento
    layers.MaxPooling1D(pool_size=2), # Riduce la dimensione (sintesi)
    layers.Dropout(0.3),

    # Secondo blocco Convoluzionale (Cerca pattern più complessi)
    layers.Conv1D(filters=64, kernel_size=3, activation='relu', padding='same'),
    layers.BatchNormalization(),
    
    # Global Average Pooling: Invece di Flatten, fa la media di tutto.
    # È ottimo per ridurre l'overfitting sui dataset piccoli.
    layers.GlobalAveragePooling1D(),
    
    # Classificatore Finale
    layers.Dropout(0.5), 
    layers.Dense(16, activation='relu', kernel_regularizer=regularizers.l2(0.01)),
    layers.Dense(1, activation='sigmoid')
])

# Compilazione
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# --- 2. CONFIGURAZIONE CALLBACKS ---
callbacks_list = [
    # Salva solo se l'accuracy sul test migliora
    ModelCheckpoint('best_model_cnn.keras', monitor='val_accuracy', save_best_only=True, mode='max', verbose=1),
    
    # Se la loss peggiora per 10 epoche, ferma tutto e torna alla versione migliore
    EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True, verbose=1),
    
    # Se non migliori per 5 epoche, riduci il learning rate
    ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=5, min_lr=1e-5, verbose=1)
]

# --- 3. ADDESTRAMENTO (Training) ---
print("\nInizio Addestramento CNN...")
history = model.fit(
    X_train, y_train,
    epochs=100, # Mettiamo 100, tanto EarlyStopping fermerà prima
    batch_size=4, # Piccolo batch size per dataset piccoli
    validation_data=(X_test, y_test),
    callbacks=callbacks_list,
    verbose=1
)

# --- 4. VALUTAZIONE FINALE E GRAFICI ---
print("\n--- VALUTAZIONE RISULTATI ---")

# Carichiamo il file migliore salvato durante il training
best_model = models.load_model('best_model_cnn.keras')

# Predizioni
y_pred_prob = best_model.predict(X_test)
y_pred = (y_pred_prob > 0.5).astype(int)

# 1. Accuracy
test_loss, test_acc = best_model.evaluate(X_test, y_test, verbose=0)
print(f"\n ACCURACY MIGLIORE SUL TEST: {test_acc*100:.2f}%")

# 2. Grafici Loss/Accuracy
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.title('Curve di Errore (Loss)')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'], label='Train Acc')
plt.plot(history.history['val_accuracy'], label='Val Acc')
plt.title('Curve di Precisione (Accuracy)')
plt.legend()
plt.show()

# 3. Matrice di Confusione
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Greens', cbar=False,
            xticklabels=['Irregolare', 'Regolare'], yticklabels=['Irregolare', 'Regolare'])
plt.xlabel('Predizione')
plt.ylabel('Realtà')
plt.title(f'CNN 1D Confusion Matrix ({test_acc*100:.1f}%)')
plt.show()
```

```python
import cv2
import mediapipe as mp
import numpy as np

# --- CONFIGURAZIONE ---
IMAGE_PATH = 'input.jpg'  # Metti qui il nome della tua foto
THICKNESS = 2             # Spessore delle linee
CIRCLE_RADIUS = 2         # Grandezza dei pallini

# --- INIZIALIZZAZIONE MEDIAPIPE ---
mp_pose = mp.solutions.pose
mp_drawing = mp.solutions.drawing_utils
mp_drawing_styles = mp.solutions.drawing_styles

# Colori personalizzati (Opzionale: stile "Cyberpunk")
# BGR format (Blue, Green, Red)
CONNECTION_COLOR = (0, 255, 0)  # Verde Neon
LANDMARK_COLOR = (0, 255, 255)  # Giallo/Ciano

def create_skeleton_images(image_path):
    # 1. Carica l'immagine
    image = cv2.imread(image_path)
    if image is None:
        print(f"Errore: Immagine '{image_path}' non trovata.")
        return

    # Ottieni dimensioni
    h, w, c = image.shape

    # 2. Crea lo sfondo nero delle stesse dimensioni
    black_background = np.zeros((h, w, c), dtype=np.uint8)

    # 3. Inizializza il modello Pose
    with mp_pose.Pose(
        static_image_mode=True,
        model_complexity=2,
        enable_segmentation=False,
        min_detection_confidence=0.5) as pose:

        # Converti BGR a RGB per MediaPipe
        image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        results = pose.process(image_rgb)

        if not results.pose_landmarks:
            print("Nessuno scheletro rilevato nell'immagine.")
            return

        # --- DISEGNO SU SFONDO NERO  ---
        mp_drawing.draw_landmarks(
            black_background,
            results.pose_landmarks,
            mp_pose.POSE_CONNECTIONS,
            landmark_drawing_spec=mp_drawing.DrawingSpec(color=LANDMARK_COLOR, thickness=CIRCLE_RADIUS, circle_radius=CIRCLE_RADIUS),
            connection_drawing_spec=mp_drawing.DrawingSpec(color=CONNECTION_COLOR, thickness=THICKNESS)
        )

        # --- DISEGNO SU FOTO ORIGINALE (Overlay) ---
        annotated_image = image.copy()
        mp_drawing.draw_landmarks(
            annotated_image,
            results.pose_landmarks,
            mp_pose.POSE_CONNECTIONS,
            landmark_drawing_spec=mp_drawing_styles.get_default_pose_landmarks_style()
        )

        # 4. Salva i risultati
        cv2.imwrite('output_skeleton_black.jpg', black_background)
        cv2.imwrite('output_overlay.jpg', annotated_image)
        
        print("Fatto! Immagini salvate:")
        print(" - output_skeleton_black.jpg (Scheletro su nero)")
        print(" - output_overlay.jpg (Scheletro su foto)")

# Esegui la funzione
create_skeleton_images(IMAGE_PATH)
```
